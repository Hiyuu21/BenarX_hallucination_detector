```python
import pandas as pd
import numpy as np

import json
import re
import random

import spacy
from spacy.lang.en.stop_words import STOP_WORDS

from sklearn.model_selection import GroupKFold, cross_val_predict
from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score, average_precision_score
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

import matplotlib.pyplot as plt
import seaborn as sns
```

## Importing RAGTruth dataset and parse the json data into dataframes
1. Filtering the data to keep only the 'good quality' responses.
2. Splitting everything into train and test sets based on the original splits.
3. Merging the generated model responses back together with their original reference context.
4. Randomly grabbing 50 unique source IDs per task from the training data to build a clean validation (dev) set, which prevents data leakage.
5. Cleaning up the final sets by dropping the old prompts, sorting the hallucination labels into specific tags, and cleanly parsing out the reference texts and questions.

```python
# R = Responses, S = Sources
RPATH = 'C:/Users/jinxl/Downloads/RAG Hallucination Detection/detector/RAGTruth/dataset/response.jsonl' 
SPATH= 'C:/Users/jinxl/Downloads/RAG Hallucination Detection/detector/RAGTruth/dataset/source_info.jsonl'

responses = pd.read_json(RPATH, lines=True)
sources = pd.read_json(SPATH, lines=True)

def get_data(data):
    to_save = []
    for idx, row in data.iterrows():
        d = row.to_dict()
        d.pop('prompt')
        d['type'] = 'response'
        label = sorted(d['labels'], key=lambda x: x['start'])
        format_label = {'baseless info': [], 'conflict': []}
        for l in label:
            if l['label_type'].lower().find('baseless')>=0:
                format_label['baseless info'].append(l['text'])
            else:
                format_label['conflict'].append(l['text'])
        d['format_label'] = format_label
        if row['task_type']=='QA':
            d['reference'] = row['source_info']['passages']
            d['question'] = row['source_info']['question']
        elif row['task_type']=='Summary':
            d['reference'] = row['source_info']
        else:
            d['reference'] = f"{row['source_info']}"
        to_save.append(d) 
    return to_save

train_responses = responses[(responses['split']=='train') & (responses['quality']=='good')]
merged_train = pd.merge(train_responses, sources, on = 'source_id', how = 'inner')
print("Merged columns: ", merged_train.columns.to_list())
print(f"Total valid training rows: {merged_train.shape[0]}\n")

test_responses = responses[(responses['split']=='test') & (responses['quality']=='good')]
merged_test = pd.merge(test_responses, sources, on = 'source_id', how = 'inner')
print("Merged columns: ", merged_test.columns.to_list())
print(f"Total valid testing rows: {merged_test.shape[0]}\n")

#sample_label = merged_train['labels'].iloc[2]
#print("Sample Label Structure:")
#print(json.dumps(sample_label, indent=4))
#print("Total usable rows: ", merged_train.shape[0] + merged_test.shape[0])

dev_source_id=[]
for task in ['QA', 'Summary', 'Data2txt']:
    source_ids = merged_train[merged_train['task_type']==task]['source_id'].unique().tolist()
    dev_source_id.extend(random.sample(source_ids, 50))

train = merged_train[~merged_train['source_id'].isin(dev_source_id)].reset_index(drop=True)
dev = merged_train[merged_train['source_id'].isin(dev_source_id)]
print(dev['task_type'].value_counts())
train['fold'] = -1
dev['fold']=-1
train = get_data(train)
dev = get_data(dev)
with open(f'./train.jsonl', 'w') as f:
    for d in train:
        f.write(json.dumps(d)+"\n")

with open(f'./dev.jsonl', 'w') as f:
    for d in dev:
        f.write(json.dumps(d)+"\n")

test = get_data(merged_test)
with open(f'./test.jsonl', 'w') as f:
    for d in test:
        f.write(json.dumps(d)+"\n")
```

## Chunking the texts into sentences - spacy

```python
def chunk_into_sentences(text):
    doc = nlp(text)
    # Extract text from the generator, filtering out purely empty lines
    sentences = [sent.text.strip() for sent in doc.sents if sent.text.strip()]
    return sentences

nlp = spacy.load("en_core_web_sm")

# A manual tracking list since tqdm is not working here
chunked_responses = []
total_rows = len(merged_train)

print(f"Starting chunking for {total_rows} rows...")

for idx, response in enumerate(merged_train['response']):
    chunked_responses.append(chunk_into_sentences(response))
    
    # Print an update every 500 rows
    if (idx + 1) % 2000 == 0:
        print(f"Processed {idx + 1} / {total_rows} rows...")
# Chunk the huge blop of response into sentences, and break each sentences into one row

clean_train_df = pd.DataFrame(get_data(merged_train))
clean_train_df['sentence_chunks'] = clean_train_df['response'].progress_apply(chunk_into_sentences)
exploded_train = clean_train_df.explode('sentence_chunks').dropna(subset=['sentence_chunks']).reset_index(drop=True)
print("Chunking done.")
```

```python
# sentence_length: The raw word count of the sentence_chunks. 
# word_overlap_ratio: The percentage of words in the generated sentence that also appear in the original reference text.
# unmatched_numbers_count: The number of numeric digits in the generated sentence that do not exist anywhere in the reference text. 

exploded_train['sentence_length'] = exploded_train['sentence_chunks'].str.split().str.len()

def calculate_overlap(row):
    sentence = str(row['sentence_chunks']).lower()
    reference = str(row['reference']).lower()

    sentence_set = set(sentence.split())
    reference_set = set(reference.split())
    intersection = sentence_set & reference_set
    
    return len(intersection)/len(sentence_set) if len(sentence_set) != 0 else 0.0

exploded_train['word_overlap_ratio'] = exploded_train.apply(calculate_overlap, axis=1)

def count_unmatched_numbers(row):
    sentence = str(row['sentence_chunks'])
    reference = str(row['reference'])
    
    # Find all number digits in both strings
    sentence_nums = re.findall(r'\d+', sentence)
    reference_nums = set(re.findall(r'\d+', reference))

    counter = 0
    for num in sentence_nums:
        if num not in reference_nums:
            counter += 1
    return counter

# Apply the function row-by-row (axis=1)
exploded_train['unmatched_numbers_count'] = exploded_train.apply(count_unmatched_numbers, axis=1)

print(exploded_train[['sentence_length', 'word_overlap_ratio', 'unmatched_numbers_count']].describe())
```

## Model Pipeline & Validation

```python
def tag_hallucination(row):
    sentence = str(row['sentence_chunks'])
    bad_strings = row['format_label']['baseless info'] + row['format_label']['conflict']

    for bad_str in bad_strings:
        if bad_str in sentence or sentence in bad_str:
            return 1
    return 0
        
exploded_train['is_hallucination'] = exploded_train.apply(tag_hallucination, axis=1)

feature_cols = ['sentence_length', 'word_overlap_ratio', 'unmatched_numbers_count']
X = exploded_train[feature_cols]
Y = exploded_train['is_hallucination']
groups = exploded_train['source_id']

gkf = GroupKFold(n_splits=5)
rf_model = RandomForestClassifier(class_weight = 'balanced', random_state=2024)

probabilities = np.zeros(len(X))
targets = np.zeros(len(X))

for fold, (train_idx, val_idx) in enumerate(gkf.split(X, Y, groups=groups)):
    X_train_fold = X.iloc[train_idx]
    X_val_fold = X.iloc[val_idx]
    Y_train_fold = Y.iloc[train_idx]
    Y_val_fold = Y.iloc[val_idx]
    print(f"Fold {fold} - Train rows: {len(X_train_fold)} | Val rows: {len(X_val_fold)}")
    
    rf_model.fit(X_train_fold, Y_train_fold)
    probabilities[val_idx] = rf_model.predict_proba(X_val_fold)[:,1]
    targets[val_idx] = Y_val_fold

best_threshold = 0.5
best_f1 = 0

for i in np.arange(0.01,1.0,0.01):
    predictions = (probabilities >= i).astype(int)
    current_f1 = f1_score(targets,predictions,zero_division=0)
    if current_f1 > best_f1:
        best_f1 = current_f1
        best_threshold = i

print(f"\nOptimal Threshold: {best_threshold:.2f}")
print(f"Best Out-of-Fold F1-Score: {best_f1:.4f}")
```

```python
# 1. Zip the features and their importance scores into a DataFrame
importance_df = pd.DataFrame({
    'Feature': feature_cols,
    'Importance': rf_model.feature_importances_
})

# 2. Sort them so the most important feature is at the top
importance_df = importance_df.sort_values(by='Importance', ascending=False)

# 3. Build the plot
plt.figure(figsize=(8, 5))
# palette='mako' gives it a sleek, modern tech color scheme
sns.barplot(x='Importance', y='Feature', data=importance_df, palette='mako', hue='Feature', legend=False)

# 4. Add formatting
plt.title('Random Forest Feature Importances: Hallucination Detection')
plt.xlabel('Gini Importance Score')
plt.ylabel('')
plt.tight_layout()

# 5. Render the chart
plt.show()
```

```python
from sklearn.metrics import precision_score, recall_score, roc_auc_score, confusion_matrix

# ---------------------------------------------------------
# STEP 1: Process the Test Set
# You must run merged_test through the exact same 
# chunking and feature extraction functions you built in Phase 2.

clean_test_df = pd.DataFrame(get_data(merged_test))
clean_test_df['sentence_chunks'] = clean_test_df['response'].progress_apply(chunk_into_sentences)
exploded_test = clean_test_df.explode('sentence_chunks').dropna(subset=['sentence_chunks']).reset_index(drop=True)
exploded_test['sentence_length'] = exploded_test['sentence_chunks'].str.split().str.len()
exploded_test['word_overlap_ratio'] = exploded_test.apply(calculate_overlap, axis=1)
exploded_test['unmatched_numbers_count'] = exploded_test.apply(count_unmatched_numbers, axis=1)
exploded_test['is_hallucination'] = exploded_test.apply(tag_hallucination, axis=1)
# ---------------------------------------------------------

# Define the test matrices
X_test = exploded_test[feature_cols]
Y_test = exploded_test['is_hallucination']

# STEP 2: Train the model on the FULL training dataset
print("Training Random Forest on the complete training set...")
rf_model.fit(X, Y)

# STEP 3: Generate probabilities on the unseen test set
print("Generating test set probabilities...")
test_probs = rf_model.predict_proba(X_test)[:, 1]

# STEP 4: Apply your optimal threshold (e.g., 0.55 from your CV phase)
optimal_threshold = best_threshold 
test_predictions = (test_probs >= optimal_threshold).astype(int)

# STEP 5: Calculate comprehensive metrics
test_precision = precision_score(Y_test, test_predictions, zero_division=0)
test_recall = recall_score(Y_test, test_predictions, zero_division=0)
test_f1 = f1_score(Y_test, test_predictions, zero_division=0)
test_auc = roc_auc_score(Y_test, test_probs)

print("\n=== FINAL TEST SET EVALUATION ===")
print(f"ROC-AUC Score: {test_auc:.4f}")
print(f"Precision:     {test_precision:.4f}")
print(f"Recall:        {test_recall:.4f}")
print(f"F1-Score:      {test_f1:.4f}")

print("\nConfusion Matrix:")
print("                 Predicted Factual | Predicted Hallucination")
matrix = confusion_matrix(Y_test, test_predictions)
print(f"Actual Factual:       {matrix[0][0]:<10d}|{matrix[0][1]:<10d}")
print(f"Actual Hallucination: {matrix[1][0]:<10d}|{matrix[1][1]:<10d}")
```

```python
import spacy
from tqdm import tqdm # Let's use this to see a progress bar so you know it's not frozen

# 1. Drop the non-"good" rows
print(f"Rows before cleaning: {len(merged_df)}")
clean_df = merged_df[merged_df['quality'] == 'good'].copy()
print(f"Rows after cleaning: {len(clean_df)}")

# 2. Load spaCy
print("\nLoading spaCy...")
nlp = spacy.load("en_core_web_sm")

# 3. Function to split sentences and apply overlap logic
def extract_sentences(row):
    text = row['response']
    labels = row['labels']
    
    doc = nlp(text)
    sentence_records = []
    
    for sent in doc.sents:
        sent_start = sent.start_char
        sent_end = sent.end_char
        is_hallu = 0
        
        # Check overlap: start1 < end2 AND start2 < end1
        if isinstance(labels, list):
            for lab in labels:
                lab_start = lab['start']
                lab_end = lab['end']
                
                if sent_start < lab_end and lab_start < sent_end:
                    is_hallu = 1
                    break # One overlap makes the whole sentence a hallucination
                    
        sentence_records.append({
            'source_id': row['source_id'],
            'split': row['split'],
            'task_type': row['task_type'],
            'model': row['model'],
            'sentence': sent.text.strip(),
            'start': sent_start,
            'end': sent_end,
            'label': is_hallu,
            # We need the original source text for feature engineering in Phase 4!
            'source_text': row['source'] 
        })
        
    return sentence_records

# 4. Apply function and flatten into a new DataFrame
print("Processing sentences (this will take 1-2 minutes)...")
tqdm.pandas()
all_sentences = clean_df.progress_apply(extract_sentences, axis=1).tolist()

# Flatten the list of lists
flat_sentences = [item for sublist in all_sentences for item in sublist]
df_sentences = pd.DataFrame(flat_sentences)

print(f"\nDone! Created {len(df_sentences)} sentences.")
print(f"Sentence-level hallucination rate: {(df_sentences['label'].mean() * 100):.1f}%")

# Let's peek at one hallucinated sentence to verify
hallucinated_sample = df_sentences[df_sentences['label'] == 1].iloc[0]
print("\n--- Verification ---")
print("Sentence:", hallucinated_sample['sentence'])
print("Label:", hallucinated_sample['label'])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from spacy.lang.en.stop_words import STOP_WORDS
import re
import numpy as np
from tqdm import tqdm

# 1. Quick fix: Replace the bogus 'source' text with the real 'source_info' text
# This saves us from having to re-run the 5-minute spaCy sentence splitting!
source_mapping = dict(zip(clean_df['source_id'], clean_df['source_info'].astype(str)))
df_sentences['source_text'] = df_sentences['source_id'].map(source_mapping)

# STEP 6: Chunk the sources (using the correct text)
print("Chunking unique sources with ACTUAL text...")
unique_sources = clean_df.drop_duplicates('source_id')[['source_id', 'source_info', 'split']].copy()

def chunk_text(text):
    text = str(text)
    raw_chunks = re.split(r'\n|\. ', text)
    valid_chunks = [c.strip() for c in raw_chunks if len(c.split()) > 2]
    if len(valid_chunks) == 0 and len(text.strip()) > 0:
        return [text.strip()]
    return valid_chunks

unique_sources['chunks'] = unique_sources['source_info'].apply(chunk_text)
source_chunks_dict = dict(zip(unique_sources['source_id'], unique_sources['chunks']))

# Fit TF-IDF on TRAIN sources only
train_chunks = unique_sources[unique_sources['split'] == 'train']['chunks'].explode().dropna().tolist()
print(f"Number of training chunks extracted: {len(train_chunks)}")

vectorizer = TfidfVectorizer(stop_words='english')
vectorizer.fit(train_chunks)
print(f"Vocabulary size: {len(vectorizer.vocabulary_)}")

# Pre-compute TF-IDF vectors
print("Pre-computing TF-IDF vectors...")
source_tfidf_dict = {
    sid: vectorizer.transform(chunks) if chunks else None 
    for sid, chunks in source_chunks_dict.items()
}

# STEP 7: Feature Engineering (Re-run)
print("Computing features for sentences (this will take 1-2 minutes)...")

def compute_features(row):
    sent = str(row['sentence'])
    sid = row['source_id']
    source_text = str(row['source_text'])
    
    # 5. Sentence length in non-stopword tokens
    words = [w for w in re.findall(r'\b\w+\b', sent.lower()) if w not in STOP_WORDS]
    length = len(words)
    if length == 0:
        return pd.Series([0.0, 0.0, 0.0, 0.0])
        
    # 2. Fraction of non-stopwords that appear in the source
    source_words = set(re.findall(r'\b\w+\b', source_text.lower()))
    words_in_source = sum(1 for w in words if w in source_words)
    frac_words = words_in_source / length
    
    # 4. Fraction of numbers NOT found in the source
    nums = set(re.findall(r'\b\d+\b', sent))
    if not nums:
        frac_num_unmatched = 0.0
    else:
        source_nums = set(re.findall(r'\b\d+\b', source_text))
        unmatched = sum(1 for n in nums if n not in source_nums)
        frac_num_unmatched = unmatched / len(nums)
        
    # 1. Max TF-IDF similarity
    max_sim = 0.0
    source_tfidf = source_tfidf_dict.get(sid)
    if source_tfidf is not None and source_tfidf.shape[0] > 0:
        try:
            sent_tfidf = vectorizer.transform([sent])
            if sent_tfidf.nnz > 0: 
                sims = cosine_similarity(sent_tfidf, source_tfidf)
                max_sim = sims.max()
        except:
            pass 
        
    return pd.Series([max_sim, frac_words, frac_num_unmatched, length])

# Apply features
tqdm.pandas()
df_sentences[['feat_max_tfidf', 'feat_word_overlap', 'feat_num_unmatched', 'feat_len']] = \
    df_sentences.progress_apply(compute_features, axis=1)

print("\nFeature extraction complete! Summary stats:")
print(df_sentences[['feat_max_tfidf', 'feat_word_overlap', 'feat_num_unmatched', 'feat_len']].describe())
```

```python
import numpy as np
from sklearn.model_selection import GroupKFold, cross_val_predict
from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score, average_precision_score
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

# 1. Split into Train and Test
train_df = df_sentences[df_sentences['split'] == 'train'].copy()
test_df = df_sentences[df_sentences['split'] == 'test'].copy()

features = ['feat_max_tfidf', 'feat_word_overlap', 'feat_num_unmatched', 'feat_len']
X_train = train_df[features]
y_train = train_df['label']
groups_train = train_df['source_id']

X_test = test_df[features]
y_test = test_df['label']

print(f"Training on {len(X_train)} sentences, Testing on {len(X_test)} sentences.")

# Define models
models = {
    "1. Always Not Hallucinated": DummyClassifier(strategy='constant', constant=0),
    "3. Logistic Regression": make_pipeline(StandardScaler(), LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42)),
    "4. Shallow Decision Tree": DecisionTreeClassifier(max_depth=3, class_weight='balanced', random_state=42),
    "5. Random Forest": RandomForestClassifier(n_estimators=100, class_weight='balanced', random_state=42, n_jobs=-1)
}

# Helper function to find best threshold using Out-Of-Fold predictions
def find_best_threshold(y_true, y_probs):
    best_thresh = 0.5
    best_f1 = 0
    # Test thresholds from 0.1 to 0.9
    for thresh in np.arange(0.1, 0.9, 0.05):
        preds = (y_probs >= thresh).astype(int)
        f1 = f1_score(y_true, preds, zero_division=0)
        if f1 > best_f1:
            best_f1 = f1
            best_thresh = thresh
    return best_thresh

results = []

# Train and Evaluate Models
print("Training models and finding optimal thresholds via Group K-Fold...")
gkf = GroupKFold(n_splits=5)

for name, model in models.items():
    print(f"Evaluating {name}...")
    
    # Setup for Rule-Based model (Model 2) which doesn't fit the standard sklearn mold
    if name == "1. Always Not Hallucinated":
        model.fit(X_train, y_train)
        test_preds = model.predict(X_test)
        test_probs = np.zeros(len(X_test)) # Dummy probs
    else:
        # Get Out-Of-Fold predictions to find the best threshold safely without touching the test set
        oof_probs = cross_val_predict(model, X_train, y_train, groups=groups_train, cv=gkf, method='predict_proba', n_jobs=-1)[:, 1]
        best_thresh = find_best_threshold(y_train, oof_probs)
        
        # Now fit on the whole train set and predict on the test set
        model.fit(X_train, y_train)
        test_probs = model.predict_proba(X_test)[:, 1]
        test_preds = (test_probs >= best_thresh).astype(int)
        print(f"  -> Selected Threshold: {best_thresh:.2f}")

    # Evaluate on Test Set
    try:
        auroc = roc_auc_score(y_test, test_probs)
        pr_auc = average_precision_score(y_test, test_probs)
    except ValueError:
        auroc, pr_auc = 0.5, 0.0 # Fallback for DummyClassifier
        
    results.append({
        "Model": name,
        "Precision": precision_score(y_test, test_preds, zero_division=0),
        "Recall": recall_score(y_test, test_preds, zero_division=0),
        "F1 (Positive)": f1_score(y_test, test_preds, zero_division=0),
        "F1 (Macro)": f1_score(y_test, test_preds, average='macro', zero_division=0),
        "AUROC": auroc,
        "PR-AUC": pr_auc
    })

# Add Model 2: Single Feature Rule (e.g., Word Overlap < 0.3 means Hallucinated)
print("Evaluating 2. Single Feature Rule (Overlap < 0.3)...")
rule_test_preds = (X_test['feat_word_overlap'] < 0.3).astype(int)
results.append({
    "Model": "2. Single Feature Rule (Overlap < 0.3)",
    "Precision": precision_score(y_test, rule_test_preds, zero_division=0),
    "Recall": recall_score(y_test, rule_test_preds, zero_division=0),
    "F1 (Positive)": f1_score(y_test, rule_test_preds, zero_division=0),
    "F1 (Macro)": f1_score(y_test, rule_test_preds, average='macro', zero_division=0),
    "AUROC": roc_auc_score(y_test, 1 - X_test['feat_word_overlap']), # Inverse because lower overlap = higher chance of hallucination
    "PR-AUC": average_precision_score(y_test, 1 - X_test['feat_word_overlap'])
})

# Display Results sorted by Model Name
results_df = pd.DataFrame(results).sort_values('Model').set_index('Model')
print("\n--- FINAL TEST SET RESULTS ---")
print(results_df.round(3).to_string())
```
